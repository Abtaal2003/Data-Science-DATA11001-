## Text generation example

This is a brief example of how to run text generation with a causal language model and `pipeline`.

Install [transformers](https://huggingface.co/docs/transformers/index) python package. This will be used to load the model and tokenizer and to run generation.

```python
!pip install --quiet transformers
```

Import the `AutoTokenizer`, `AutoModelForCausalLM`, and `pipeline` classes. The first two support loading tokenizers and generative models from the [Hugging Face repository](https://huggingface.co/models), and the last wraps a tokenizer and a model for convenience.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
```

Load a generative model and its tokenizer. You can substitute any other generative model name here (e.g. [other TurkuNLP GPT-3 models](https://huggingface.co/models?sort=downloads&search=turkunlp%2Fgpt3)), but note that Colab may have issues running larger models. 

```python
MODEL_NAME = 'TurkuNLP/gpt3-finnish-large'

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)
```

Instantiate a text generation pipeline using the tokenizer and model.

```python
pipe = pipeline(
    'text-generation',
    model=model,
    tokenizer=tokenizer,
    device=model.device
)
```

We can now call the pipeline with a text prompt; it will take care of tokenizing, encoding, generation, and decoding:

```python
output = pipe('Terve, miten menee?', max_new_tokens=25)

print(output)
```

Just print the text

```python
print(output[0]['generated_text'])
```

We can also call the pipeline with any arguments that the model `generate` function supports. For details on text generation using `transformers`, see e.g. [this tutorial](https://huggingface.co/blog/how-to-generate).

Example with sampling and a high `temperature` parameter to generate more chaotic output:

```python
output = pipe(
    'Terve, miten menee?',
    do_sample=True,
    temperature=10.0,
    max_new_tokens=25
)

print(output[0]['generated_text'])
```

## Exercise 1

Your task is to assess whether a generative model succeeds in the following tasks in zero-shot, one-shot, and two-shot settings:

- binary sentiment classification (positive / negative)

- person name recognition

- two-digit addition (e.g. 11 + 22 = 33)

For example, for assessing whether a generative model can name capital cities, we could use the following prompts:

- zero-shot:

- two-shot:
	>"""\

	>
	>Question: What is the capital of Sweden?\
	>Answer: Stockholm
	>
	>Question: What is the capital of Denmark?\
	>Answer: Copenhagen
	>

	>Answer:\
	>"""

You can do the tasks either in English or Finnish and use a generative model of your choice from the Hugging Face models repository, for example the following models:

- English: `gpt2-large`
- Finnish: `TurkuNLP/gpt3-finnish-large`

You can either come up with your own instructions for the tasks or use the following:

- English:
	- binary sentiment classification: "Do the following texts express a positive or negative sentiment?"
	- person name recognition: "List the person names occurring in the following texts."
	- two-digit addition: "This is a first grade math exam."
- Finnish:
	- binary sentiment classification: "Ilmaisevatko seuraavat tekstit positiivista vai negatiivista tunnetta?"
	- person name recognition: "Listaa seuraavissa teksteissä mainitut henkilönnimet."
	- two-digit addition: "Tämä on ensimmäisen luokan matematiikan koe."

Come up with at least two test cases for each of the three tasks, and come up with your own one- and two-shot examples.

```python
# Use this cell for your code

# I did the tasks in English with gpt2-large
MODEL_NAME = 'gpt2-large'

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)

pipe = pipeline(
    'text-generation',
    model=model,
    tokenizer=tokenizer,
    device=model.device
)


def make_prompt(instruction, label, examples, question):
    prompt = instruction + '\n\n'
    for q, a in examples:
        prompt += label + ': ' + q + '\nAnswer: ' + a + '\n\n'
    prompt += label + ': ' + question + '\nAnswer:'
    return prompt


def run_task(name, instruction, label, examples, tests):
    print('=====', name, '=====')
    for k in [0, 1, 2]:
        print('---', k, 'shot ---')
        for question, correct in tests:
            prompt = make_prompt(instruction, label, examples[:k], question)
            output = pipe(prompt, max_new_tokens=10, do_sample=False, return_full_text=False)
            # only keep the first line of the generated answer
            answer = output[0]['generated_text'].strip().split('\n')[0]
            print(question, '->', answer, '| correct answer:', correct)
    print()


# Task 1: sentiment classification
sentiment_examples = [
    ('I really enjoyed this book, it was great.', 'positive'),
    ('This phone broke after one day, terrible.', 'negative')
]
sentiment_tests = [
    ('The food was delicious and the staff were very friendly.', 'positive'),
    ('I waited two hours and the package never arrived.', 'negative')
]
run_task('Sentiment', 'Do the following texts express a positive or negative sentiment?',
         'Text', sentiment_examples, sentiment_tests)

# Task 2: person name recognition
name_examples = [
    ('Mikko and Laura went to the cinema.', 'Mikko, Laura'),
    ('The book was a gift from David Smith.', 'David Smith')
]
name_tests = [
    ('Yesterday Anna met Pekka at the library.', 'Anna, Pekka'),
    ('The report was written by Maria Virtanen and checked by John.', 'Maria Virtanen, John')
]
run_task('Person names', 'List the person names occurring in the following texts.',
         'Text', name_examples, name_tests)

# Task 3: two-digit addition
math_examples = [
    ('11 + 22', '33'),
    ('26 + 15', '41')
]
math_tests = [
    ('34 + 25', '59'),
    ('47 + 38', '85')
]
run_task('Addition', 'This is a first grade math exam.',
         'Question', math_examples, math_tests)
```

### Answers / observations

**Sentiment classification**

- Zero-shot: the model did not answer with positive/negative at all. For the first text it just repeated the text and for the second it wrote "I am very disappointed.", so it kind of continued the review instead of classifying it. Failed both.
- One-shot: it answered "negative" for both texts, so the negative one was correct but the positive one was wrong. The single example showed the answer format but the model did not really understand the task.
- Two-shot: both correct (positive and negative). With one positive and one negative example the model got both test cases right.

**Person name recognition**

- Zero-shot: failed, it wrote a sentence ("Pekka is a friend of Anna's.") or repeated the text instead of listing names.
- One-shot: first case fully correct (Anna, Pekka). Second case partly correct, it found Maria and John but dropped the surname Virtanen.
- Two-shot: first case only found Anna and missed Pekka. Second case found all names (Maria Virtanen and John), just in a slightly different format. So adding the second example did not clearly improve the results, it was about the same as one-shot.

**Two-digit addition**

- Zero-shot: failed, it just repeated the question (34 + 25).
- One-shot: failed, it answered with the first number of the question (34, 47).
- Two-shot: gave numbers in the right format but both were wrong (49 instead of 59, 60 instead of 85).
- The model could not do addition in any setting.

**Summary**

Zero-shot did not work for any task because the model just continues the text and does not follow the instruction. The examples mostly help the model understand the answer format. With two examples sentiment worked well and name recognition worked partly, but addition did not work at all, so gpt2-large seems to copy patterns rather than actually calculate.
